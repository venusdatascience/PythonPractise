```python
#regular expression:It is use to find pattern in text

import re #import remodule

#functions of regular expression regex

#re.search:look the pattern anywhere in the text

txt ="pyhton is a programing language"

result=(re.search("is",txt))
print(result)
result.group() # group():what was found
result.span() #at what index it was fount


```

```python
import re

txt = """
Hello, my name is Venus and I am learning Python and Data Science.

You can contact me at venus@gmail.com or venus123@yahoo.com.
My phone numbers are 9841234567 and 9812345678.

My website is https://www.github.com and I also visit https://www.google.com.

I am learning #Python, #DataScience and #MachineLearning.

I scored 85 in Python, 90 in SQL and 88 in Data Science.

Some example passwords are Python123 and DataScience2026.
"""
```

```python
#email extraction
pattern = r"[\w.]+@[\w.]+\.\w+" # r represent any raw pattern [] set of charcter\w(any word that contain a-z charcter 0-9 digit and underscore)

#.(any charcter except new line) +(one or more occurance ) \.(represent special dot) \w+(one or moreoccurance containg a=z,0-9)

print(re.findall(pattern,txt))
```

```python
pattern1 = r"\b\d{10}\b"


phn = re.findall(pattern1,txt)
print(phn)

for i in phn:
    if re.match( r"[0-9]{10}", i):
        print("Valid phone number")
        
    else:
        print("invalid number")


print(re.findall(pattern1,txt))



```

```python
#password

password1 ="Python123"
pattern2 = r"^(?=.*[A-Z])(?=.*[a-z])(?=.*\d).{8,}$"

if re.match(pattern2, password1):
    print("Valid password")
else:
    print("Invalid password")



```

```python
#url extraction
pattern3 = r"https?://[^\s]+"

urls = re.findall(pattern3, txt)

print(urls)
```

```python
#finding hashtag

pattern4 = r"#\w+"

hastag = re.findall(pattern4,txt)
print(hastag)
```

```python
#extracting number from text
pattern5 = r"\d+"

numtxt = re.findall(pattern5,txt)
print(numtxt)
```

```python
import re

with open("data.txt", "r") as f:
    data = f.read()

    print(data)
    
```

```python
type(data)
```

```python
email_pattern = r"[\w.]+@[\w.]+\.\w+"
phone_pattern = r"[0-9]{10}"
url_pattern = r"https?://[^\s]+"

emails = re.findall(email_pattern, data)
phones = re.findall(phone_pattern, data)
urls = re.findall(url_pattern, data)

print("Emails:", emails)
print("Phones:", phones)
print("URLs:", urls)
```

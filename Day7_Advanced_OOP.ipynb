```python
#adding init function to child class

class Fruit:
    def __init__(self, name, taste):
        self.name = name
        self.taste = taste

    def describe(self):
        print(self.name)
        print(self.taste)


class NewFruit(Fruit):
    def __init__(self, name, taste, colour):#adding own init function
        Fruit.__init__(self, name, taste)#inherinting init  function from parent if we do not want to overide parent 
        self.colour = colour

    def show_colour(self): #adding function to child clas
        print(self.colour)


x1 = NewFruit("banana", "sweet", "yellow")

x1.describe()
x1.show_colour()
```

```python
#multiple inheritance
class Mother:
    def __init__(self,name,skill):
        self.name= name
        self.skill=skill

    def info(self):
        print(self.name)
        print(f"i have these skill {self.skill}")


class Father:
    def __init__(self,nature):
        self.nature=nature
        print(f"i m {self.nature} in nature")

class Child(Mother,Father):
    
    def __init__(self,name,skill,nature):
        Mother.__init__(self,name,skill)
        Father.__init__(self,nature)

    def info(self):
        print(self.name)
        print(self.skill)
        print(self.nature)

        
  


x1 = Child("tina","communication","sweet")
print(x1.info())
    

```

```python
#

class Teacher:
    def __init__(self,name,subject):
        self.name=name
        self.subject=subject


class Skills:
    def __init__(self,skill):
        self.skill=skill


class Student(Teacher,Skills):
    def __init__(self,name,subject,skill):
       Teacher.__init__(self,name,subject)
       Skills.__init__(self,skill)

    def info(self):
        print(self.name)
        print(self.subject)
        print(self.skill)

student1 = Student("ram","science","java")
print(student1.info())
    
```

```python
#Abstract classAn abstract class is a class that is used as a blueprint for child classes and contains one or more abstract methods that child classes are required to implement.

from abc import ABC, abstractmethod

class Employee(ABC):

    @abstractmethod
    def work(self): #it shows work() method should be there in child classes
        pass


class Developer(Employee):

    def work(self):
        print("Developer writes code")


class Teacher(Employee):

    def work(self):
        print("Teacher teaches")



d = Developer()
t = Teacher()

d.work()
t.work()

```

```python
#__str__:it controls opuput

class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self): #if we doesnot use __strt__then after creating object and printing it it doesnot shows anything so __str__controls output
        return self.name


student = Student("Venus")
print(student)
```

```python
#__len__() — Control len() length of character
class Student:

    def __init__(self, name):
        self.name = name

    def __len__(self): #when object is made it directly give length of charcter wihout it it len() function  doesnot give length 
        return len(self.name)



student = Student("Venus")

print(len(student))
```

```python
#composition :it means one class conatin and object of anoter class

class Engine:
    def start(self):
        print("Engine starts")


class Car:
    def __init__(self):
        self.engine = Engine() #contain object of engine class

    def start_car(self):
        self.engine.start()
        print("Car starts")


car = Car()
car.start_car()



```

```python
#vehicle management system

class Vehicle:
    def start(self):
        print("Vehicle starts")


# Inheritance + Polymorphism
class Car(Vehicle):
    def start(self):
        print("Car starts")


class Bike(Vehicle):
    def start(self):
        print("Bike starts")


class Truck(Vehicle):
    def start(self):
        print("Truck starts")


# Objects
car = Car()
bike = Bike()
truck = Truck()


# Polymorphism
car.start()
bike.start()
truck.start()
```

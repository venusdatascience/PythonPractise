```python
#Count positive and negative numbers
numbers = [10, -5, 7, -3, 8, -2, 4]
count = 0
count1 =0
for i in numbers:
    if i>1:
        count = count+1
    elif i<1:
        count1 = count1+1
        
print(f"the count o positive numbers {count}")
print(f"the count of negative numbers {count1}")
        
```

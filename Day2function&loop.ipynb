```python
#scope:it means where the variable can be accessed
#local and global variables

#local variable :which can be access only inside the function

def place():
    x1 = "ktm"
    return x1
place()
print(x1) #cannot be acces outside of fucntion

```

```python
#global variable:it can be access outside of function
def place():
    global x1
    x1 ="ktm"
    return x1

print(place())
print(x1) #here variable is global variable so it can be access outside of function 
```

```python
#number gusing
number = 2

def new_number(x):
    while True:
        if x == number:
            return "your guess is correct"
        else:
            x = int(input("try again, enter a number: "))

y = int(input("enter a number: "))
print(new_number(y))
```

```python
#prime number
def prime_number(x):
    for i in range(2, x):
        if x % i == 0:
            print("not a prime number")
            return

    print("prime number")

y = int(input("enter a number"))
prime_number(y)
```

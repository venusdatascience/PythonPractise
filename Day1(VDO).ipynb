```python
#comparision operators:compares two values and gives true or false
#==,!=,>=,>=

age = 20
print(age==20) #equal to
print(age>40)
print(age!=20) #not eqal to
print(age<20)
print(age<50)
print(age>10)

```

```python
#identity operators :compares object with same values and same memory location
#is,is not

a = [1,2,3]
b = a
print(a is b)

print(a is not b) #same id
print(id(a))
print(id(b))


```

```python
#diffrence between is and ==

#== check if both have same values
# is if both variable refer to same object in memory
x = [1,2,3]
y = [1,2,3]

print(x==y)
print(x is y) # x and y refer to seprate list 

print(id(x)) #diffent id
print(id(y))


```

```python
#bit wise operators:this are use to comapre binary numbers
#and(&),or(|),xor(^),not

#& comapres two bit if same gives 1 otherwise 0
a = 4
b= 5

print(bin(a))
print(bin(b))
print("And",a & b) #both number are 1  then 1 otherwise o
print("OR", a | b) # any one is 1 then 1
print("XOR", a ^b) #if only one is 1 zero if both are 1 or zero)


```

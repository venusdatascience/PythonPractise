```python
#identity operators :compares object with same values and same memory location
#is,is not

a = [1,2,3]
b = a
print(a is b)

print(a is not b) #same id
print(id(a))
print(id(b))


```

```python
#diffrence between is and ==

#== check if both have same values
# is if both variable refer to same object in memory
x = [1,2,3]
y = [1,2,3]

print(x==y)
print(x is y) # x and y refer to seprate list 

print(id(x)) #diffent id
print(id(y))


```

```python
#modifying string
#lower(),upper(),stip(),split(),replace()

city = "  kathmandu city  "
print(city.lower())
print(city.upper())
print(city.split("a")) #split() method return a list where the text between the givien separator become new list item)
print(city.split(" "))
print(city.strip()) #removes white space from begining or end of the text

print(city.replace("city","place")) #replace a string with another string




```

```python
#string methods .capitalize()	Capitalizes first character	"hello".capitalize() → Hello
#capitalize(),title(),count(),find(),startswith(),endswith(),isdigit(),isalpha(),isalnum(),join()

food = "pasta  23 susi"
print(food.capitalize()) #capitalize first letter of a string
print(food.title())#capitalize first letter of each word
print(food.count("a")) #counts number of letters in a string

print(food.find("t")) #find the position of a letter in string
print(food.startswith("p")) 
print(food.endswith("i"))

print(food.isdigit())
```
